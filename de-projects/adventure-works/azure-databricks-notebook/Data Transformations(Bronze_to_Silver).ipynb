# Transformations

## AdventureWorks_Calendar

```python
df_cal = df_cal.withColumn('Month',month(col('Date')))\
      .withColumn('Year',year(col('Date')))

df_cal.limit(5).display()
```

```python
df_cal.write.format('parquet').mode('overwrite').save('abfss://silver@shoaibadlsdev.dfs.core.windows.net/AdventureWorks_Calendar')
```

## AdventureWorks_Customers

```python
df_cus.limit(5).display()
```

```python
df_cus = df_cus.withColumn("Fullname",concat_ws(" ",col("Prefix"),col("FirstName"),col("LastName")))
df_cus.limit(5).display()
```

```python
df_cus.write.format('parquet').mode('overwrite').save('abfss://silver@shoaibadlsdev.dfs.core.windows.net/AdventureWorks_Customers')
```

## Product_Subcategories

## AdventureWorks_Products

```python
df_pro.limit(5).display()
```

```python
df_pro = df_pro.withColumn("ProductSKU",split(col("ProductSKU"),"-")[0])\
               .withColumn("ProductName",split(col("ProductName")," ")[0])

df_pro.limit(5).display()
      
```

```python
df_pro.write.format('parquet').mode('overwrite').save('abfss://silver@shoaibadlsdev.dfs.core.windows.net/AdventureWorks_Products')
```

## Sales

```python
df_sales.limit(5).display()
```

```python
df_sales = df_sales.withColumn("StockDate",to_timestamp(col("StockDate")))\
                  .withColumn("OrderNumber",regexp_replace(col("OrderNumber"),"S","T"))\
                  .withColumn("Multiply",col("OrderLineItem")*col("OrderQuantity"))

df_sales.limit(5).display()
```

```python
df_sales.write.format('parquet').mode('overwrite').save('abfss://silver@shoaibadlsdev.dfs.core.windows.net/AdventureWorks_Sales')
```
